```python
# 🔰 OUTLIER DETECTION & REMOVAL: IQR METHOD

# The Interquartile Range (IQR) method is robust to skewed distributions.
# It flags values outside [Q1 - 1.5*IQR, Q3 + 1.5*IQR] as outliers.

def iqr_bounds(series, k=1.5):
    """Compute lower and upper bounds using Tukey's rule (k × IQR)."""
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - k * iqr
    upper = q3 + k * iqr
    return lower, upper

# Ensure df_num exists (it is normally created earlier). If not, try to build it from `data`.
if 'df_num' not in globals():
    if 'data' not in globals():
        raise NameError("data is not defined. Run the data loading cell before this one.")
    num_cols = [
        'vehicle_age', 'km_driven', 'mileage',
        'engine', 'max_power', 'seats', 'selling_price'
    ]
    missing_cols = [c for c in num_cols if c not in data.columns]
    if missing_cols:
        raise KeyError(f"The following expected numeric columns are missing from `data`: {missing_cols}")
    df_num = data[num_cols].copy()

#  🔰 Build masks per feature
iqr_masks = []
for col in df_num.columns:
    low, high = iqr_bounds(df_num[col], k=1.5)
    mask = (df_num[col] < low) | (df_num[col] > high)
    iqr_masks.append(mask)
    print(f"{col}: bounds=({low:.2f}, {high:.2f}) | outliers={int(mask.sum())}")

#  🔰 Combine masks across all features
iqr_any_mask = pd.concat(iqr_masks, axis=1).any(axis=1)

#  🔰 Remove rows flagged as outliers
df_iqr_clean = df_num[~iqr_any_mask].copy()
print("Shape after IQR filtering:", df_iqr_clean.shape)

# 🔰 Create cleaned DataFrame df_clean = df_num[~iqr_any_mask].copy() 
# print("Shape after IQR filtering:", df_clean.shape)
```

```python

#  🔰 OUTLIER DETECTION & REMOVAL: Z-SCORE METHOD

# Z-score measures how many standard deviations a value is from the mean.
# Outliers are flagged when |z| > 3

from scipy.stats import zscore
import numpy as np

# 🔰 Compute z-scores for each feature
z_scores = df_num.apply(zscore)

# 🔰 Flag outliers where |z| > threshold
z_threshold = 3
z_outlier_mask = (np.abs(z_scores) > z_threshold)

# 🔰 Summarize outliers per feature
print("Z-score outlier counts per feature:\n", z_outlier_mask.sum())

# 🔰 Remove rows with any feature exceeding threshold
df_z_clean = df_num[~z_outlier_mask.any(axis=1)].copy()
print("Shape after Z-score filtering:", df_z_clean.shape)
```

#### 4.5 Encode categorical features

```python

# STEP 25: 🚗  ENCODING CATEGORICAL FEATURES-SUMMARY

#🔰Common Encoding Techniques

#🔰Label Encoding
#✅  Assigns each category a unique integer.
#✅ Best for ordinal features (where order matters, e.g., owner count).
#✅ Risk: For nominal features, models may misinterpret integers as ordered.

#🔰One‑Hot Encoding
#✅ Creates binary columns for each category (0/1).
#✅ Best for nominal features (no natural order, e.g., fuel, seller_type).
#✅ Expands dimensionality if many categories exist.

# 🔰Target Encoding
# ✅ Replaces categories with the mean of the target variable for that category.
# ✅ Useful for high-cardinality features (many unique categories).
# ✅ Risk of target leakage if not done carefully (e.g., within cross-validation folds).

# 🔰 Recommended Encoding Strategies:
# ✅  Since we will be mixing tree‑based and non‑tree algorithms:
# ✅ Used One‑Hot Encoding as the default .It works universally across all models.
# ✅ Linear models (Regression, Logistic Regression): avoids false ordinal relationships.
# ✅ Distance‑based models (KNN, SVM): ensures categorical variables don’t distort distance metrics.
# ✅ Tree‑based models (Random Forest, XGBoost): can handle one‑hot encoded features well.
```

```python
#  STEP 26: 🚗 ENCODING CATEGORICAL FEATURES: ONE-HOT ENCODING

from sklearn.preprocessing import OneHotEncoder
import pandas as pd

# 🔰: Load dataset
data = pd.read_csv('car_price_dataset.csv')

# 🔰: Quick check of dataset structure
# Print shape (rows, columns) and column names to confirm successful load
print("✅ Dataset loaded. Shape:", data.shape)
print()
print("✅ Columns:", data.columns)

# 🔰 Decide encoding strategy for categorical features
# ✅ All selected features are nominal (no natural order), One-Hot Encoding is appropriate
# ✅ fuel_type: Nominal =One-Hot Encoding
# ✅ seller_type: Nominal =One-Hot Encoding
# ✅ transmission_type: Nominal =One-Hot Encoding

# 🔰 Select categorical features for encoding
cat_cols = ['fuel_type', 'seller_type', 'transmission_type']

# 🔰: Apply One-Hot Encoding
# drop_first=True removes the first category to avoid multicollinearity (dummy variable trap)
data_encoded = pd.get_dummies(data, columns=cat_cols, drop_first=True)

# 🔰: Verify new dataset structure
print()
print("✅ One-Hot Encoding applied. Dataset now has these columns:")
print()
print(data_encoded.columns)
print("\n✅ New shape after encoding:", data_encoded.shape)

# 🔰 : Observations
# ✅ One-Hot Encoding created binary columns for each category (excluding the first).
# ✅ This ensures categorical features are represented without imposing false ordinal relationships.
# ✅ The dataset is now ready for modeling with both tree-based and non-tree algorithms.

# 🚗 FINAL DATASET READY FOR MODELING

# ✅ Confirm that categorical features have been successfully encoded into numeric columns
print("✅ Columns increase = categorical features now represented numerically.")
print("✅ The dataset is now ready for modeling across linear, distance‑based, and tree‑based algorithms.")

# 🔰 Before vs after comparison table
print()

# 🔰 : Show original categorical columns (before encoding)
# ✅ This displays the raw string values for the first 5 rows
print("🔰 BEFORE ENCODING (sample rows)")
print(data[['fuel_type', 'seller_type', 'transmission_type']].head(5))

print()

# 🔰 : Show encoded columns (after One-Hot Encoding)
# ✅ The regex filter selects only the new dummy columns created for these features
print("🔰 AFTER ENCODING (sample rows)")
print(data_encoded.filter(regex='fuel_type|seller_type|transmission_type').head(5))
```

```python
# STEP 34: 🚗 TRAIN THREE REGRESSION MODELS ON TOP-K FEATURES

# 🔰 STEP: 00 : FEATURE PREPARATION

#✅  Ensure Top-K feature selection pipeline has been run first to create X_corr_selected and y.
# Train and evaluate three regression models on the Top‑K correlated features:

# ✅ Linear Regression (with Ridge regularization) — interpretable baseline.
# ✅ Random Forest Regressor — non‑linear, robust to multicollinearity.
# ✅ Gradient Boosting Regressor — strong performance with careful learning rate.
# ✅  Each model uses the same train/test split and evaluation metrics (MAE, RMSE, R²) for fair comparison.

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# 🔰 Run and create the Top‑K feature selection pipeline first to ensure these variables exist:

#  ✅ df_num: numeric working dataframe
#  ✅  TARGET: set to 'selling_price'
#  ✅ X_corr_selected: DataFrame of selected Top‑K features
#  ✅ y: target series

# 🔰 Load dataset

data = pd.read_csv('car_price_dataset.csv')
TARGET = 'selling_price'

NUMERIC_FEATURES = [
    'vehicle_age','km_driven','mileage','engine','max_power','seats','selling_price'
]

# 🔰 Check and numeric subset
missing = [c for c in NUMERIC_FEATURES if c not in data.columns]
if missing:
    raise ValueError(f"Missing expected columns: {missing}")
df_num = data[NUMERIC_FEATURES].copy()

# 🔰 Correlation-based Top‑K selection
corr_matrix = df_num.corr()
corr_target = corr_matrix[TARGET].drop(TARGET)
corr_abs = corr_target.abs().sort_values(ascending=False)

K = 4
top_features = corr_abs.head(K).index.tolist()
X_corr_selected = df_num[top_features].copy()
y = df_num[TARGET].copy()

print("✅ Using features:", top_features)
```

```python
#  STEP: 35: 🚗 TRAIN-TEST SPLIT

# 🔰 ML PIPELINE: Train three regression algorithms on car price dataset

# ✅ Import required libraries
# - train_test_split: splits dataset into training and testing subsets
# - mean_absolute_error, mean_squared_error, r2_score: evaluation metrics (used later)
# - numpy & pandas: numerical and tabular data handling
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

# 🔰 Train/Test Split
# ✅ Separate data into training and testing sets to evaluate model performance fairly.
# - Training set (80%): used to fit/train the ML models
# - Testing set (20%): held back to evaluate how well the model generalizes to unseen data
# - random_state=42: ensures reproducibility (same split every run)
# - test_size=0.2: reserves 20% of the dataset for testing
X_train, X_test, y_train, y_test = train_test_split(
    X_corr_selected,  # ✅ Selected Top-K features (predictors)
    y,                # ✅ Target variable (selling_price)
    test_size=0.2,    # ✅ 20% of data for testing
    random_state=42   # ✅ Fixed seed for reproducibility
)

# ✅ Print shapes of the resulting datasets:
# - X_train: (number of training samples, number of features)
# - X_test:  (number of testing samples, number of features)
# - y_train: (number of training samples,)
# - y_test:  (number of testing samples,)
# ✅ This helps confirm that the train/test split worked correctly and that
#    the dimensions align (same number of rows for X and y in each split).

print(" ✅ Train-Test Split Complete:")
print(" ✅X_train shape:", X_train.shape)   # Training predictors (rows = samples, cols = features)
print(" ✅X_test shape:", X_test.shape)     # Testing predictors
print(" ✅y_train shape:", y_train.shape)   # Training target values
print(" ✅y_test shape:", y_test.shape)     # Testing target values

# 🔰 Summary:
# - X_train: training predictors (used to fit/train ML models)
# - X_test:  testing predictors (used to evaluate model generalization)
# - y_train: training target values (selling_price for training set)
# - y_test:  testing target values (selling_price for test set)
print()
print(" 🔰These will be used in subsequent steps to train and evaluate ML models.")
```

```python
#  STEP 36: 🚗 METRICS

# ✅ This helper function evaluates regression model predictions using three common metrics:
# ✅ MAE (Mean Absolute Error): Average absolute difference between predicted and actual values.
# ✅ RMSE (Root Mean Squared Error): Penalizes larger errors more heavily; useful for risk-sensitive tasks.
# ✅ R² (Coefficient of Determination): Proportion of variance in the target explained by the model.

def evaluate_regression(model_name, y_true, y_pred):
    # ✅ Compute Mean Absolute Error (lower is better)
    mae = mean_absolute_error(y_true, y_pred)
    
    # ✅ Compute Root Mean Squared Error (lower is better, penalizes large errors)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))

    # ✅ Compute R² score (higher is better, max = 1.0)
    r2 = r2_score(y_true, y_pred)
    
    # ✅ Print results in a stakeholder-friendly format
    print(f"\n✅  {model_name} — Performance on Test Set")
    print(f"- MAE:  {mae:,.2f}")   # Mean Absolute Error: typical error magnitude
    print(f"- RMSE: {rmse:,.2f}")  # Root Mean Squared Error: emphasizes large deviations
    print(f"- R²:   {r2:,.4f}")    # Variance explained: closer to 1 means better fit
    
    # ✅ Return results as a dictionary for easy comparison across models
    return {'model': model_name, 'MAE': mae, 'RMSE': rmse, 'R2': r2}
```

```python
#  STEP 37: 🚗 TRAIN & EVALUATE MODELS

# ==================🔰 MODEL 1: Ridge Regression (Linear model with regularization)================

# ✅ Import required libraries
# - Ridge: linear regression with L2 regularization (helps with multicollinearity and stabilizes coefficients)
# - Pipeline: allows chaining preprocessing steps with the model
# - StandardScaler: standardizes features (mean=0, variance=1) so units don’t dominate
from sklearn.linear_model import Ridge
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

# ✅ Create a pipeline for Ridge Regression
# - Step 1: Scale features (important for linear models so coefficients are comparable)
# - Step 2: Fit Ridge regression model
# - alpha=1.0: regularization strength (higher alpha = stronger penalty on large coefficients)
ridge_pipe = Pipeline(steps=[
    ('scaler', StandardScaler()),                 # normalize feature scales
    ('ridge', Ridge(alpha=1.0, random_state=42))  # apply Ridge regression
])

# ✅ Train Ridge model on training data
# - Fits the pipeline: scaler learns scaling parameters, Ridge learns regression coefficients
ridge_pipe.fit(X_train, y_train)

# ✅ Predict on test data
# - Uses the trained pipeline to scale X_test and generate predictions
ridge_pred = ridge_pipe.predict(X_test)

# ✅ Evaluate Ridge model
# - Calls evaluate_regression() to compute MAE, RMSE, and R²
ridge_metrics = evaluate_regression("Ridge Regression", y_test, ridge_pred)
print()
print("🔰 INTERPRETATION OF RIDGE REGRESSION RESULTS")

print("\n✅ Interpretation of Ridge Regression Performance:")
print("✅ The Mean Absolute Error (MAE) of 280,307.48 indicates that, on average,")
print("   the model's predictions are off by about R280k compared to actual selling prices.")
print("✅ The Root Mean Squared Error (RMSE) of 502,068.56 shows that larger errors")
print("   are penalized more heavily, meaning some predictions deviate by ~R500k.")
print("✅ The R² score of 0.6651 means the model explains about 66.5% of the variance")
print("   in car selling prices, which is a reasonably strong fit for a linear baseline model.")
```

```python
# 🔰 LOAD HELPER FUNCTION

# Defines a reusable function to evaluate regression models.
# It ensures consistent reporting of metrics across Ridge, Random Forest, and Gradient Boosting.

# ✅ Import regression metrics from scikit-learn
# - mean_absolute_error (MAE): average absolute difference between predictions and actual values
# - mean_squared_error (MSE): average squared difference; we use squared=False to get RMSE
# - r2_score (R²): proportion of variance explained by the model
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# ✅ Define helper function
def evaluate_regression(model_name, y_true, y_pred):
    """
    Evaluate regression predictions with MAE, RMSE, and R².
    
    Parameters:
    - model_name (str): name of the model being evaluated (e.g., "Random Forest (Tuned)")
    - y_true (array-like): actual target values from the test set
    - y_pred (array-like): predicted values from the model
    
    Returns:
    - dict: structured dictionary containing model name and evaluation metrics
            {
              "model": <model_name>,
              "MAE": <mean absolute error>,
              "RMSE": <root mean squared error>,
              "R2": <r-squared score>
            }
    
    Purpose:
    - Provides transparency and auditability by standardizing metric reporting.
    - Makes it easy to consolidate results across multiple models into a DataFrame.
    """
    return {
        "model": model_name,
        "MAE": mean_absolute_error(y_true, y_pred),                     # average prediction error
        "RMSE": mean_squared_error(y_true, y_pred, squared=False),      # penalizes larger errors more heavily
        "R2": r2_score(y_true, y_pred)                                  # variance explained by the model
    }
```

```python
# STEP 48: 🚗 MODEL SERIALIZATION

# ✅ Serialization is a critical step in the machine learning lifecycle.
# ✅ It ensures that trained models (including preprocessing steps) can be reused consistently
#  across environments without retraining. This supports reproducibility, auditability,
# ✅ and deployment readiness in production systems (Streamlit)

# ✅ Serialization preserves the trained pipeline (preprocessing + model) in a portable file format.
# ✅ By saving the entire pipeline, we guarantee inference uses the exact same transformations
#    applied during training (scaling, feature selection, etc.).
# ✅ joblib is preferred for scikit-learn objects because it efficiently handles large numpy arrays.

# 🔰 FIT THE MODEL

# ✅Fitting the pipeline before serialization ensures that the model is trained
# with cross-validation tuned hyperparameters, minimizing overfitting and maximizing generalization.

# ✅ Import required classes (guarded to avoid re-import errors in notebooks)
if 'GridSearchCV' not in globals():
    from sklearn.model_selection import GridSearchCV
if 'StandardScaler' not in globals():
    from sklearn.preprocessing import StandardScaler
if 'ColumnTransformer' not in globals():
    from sklearn.compose import ColumnTransformer
if 'Pipeline' not in globals():
    from sklearn.pipeline import Pipeline
if 'GradientBoostingRegressor' not in globals():
    from sklearn.ensemble import GradientBoostingRegressor

# ✅ Define feature sets
numeric_features = ['max_power', 'engine', 'mileage', 'vehicle_age']
categorical_features = []  # none present in X_train

# ✅ Define hyperparameter grid (step-prefixed for Pipeline compatibility)
if 'gbr_params' not in globals():
    gbr_params = {
        'gbr__n_estimators': [200, 300, 500],
        'gbr__learning_rate': [0.01, 0.05, 0.1],
        'gbr__max_depth': [3, 4, 5]
    }

# ✅ Preprocessor: scaling numeric features
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features)
    ])

# ✅ Pipeline: integrates preprocessing + Gradient Boosting model
model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('gbr', GradientBoostingRegressor(random_state=42))
])

# ✅ GridSearchCV: tunes hyperparameters with CV to minimize overfitting
grid = GridSearchCV(model, gbr_params, cv=3, scoring='neg_mean_squared_error', n_jobs=-1)
grid.fit(X_train, y_train)   # Fit pipeline before serialization

# 🔰 SAVE THE FITTED PIPELINE

# ✅ Saving the best estimator ensures reproducibility and portability.
import joblib
joblib.dump(grid.best_estimator_, "gradient_boosting_pipeline.pkl")
print("✅ Model pipeline saved successfully")

# 🔰 CONFIRM THE FILE EXISTS

# Verifying file existence adds robustness and auditability.
import os
print(os.path.exists("gradient_boosting_pipeline.pkl"))  # should return True

# 🔰 LOAD THE MODEL BACK

# Reloading validates that the serialized object can be restored without loss.
# This step is critical for deployment workflows (e.g., Streamlit), ensuring portability.
model_api = joblib.load("gradient_boosting_pipeline.pkl")
print("✅ Model pipeline loaded successfully and ready for deployment")
```

```python

# STEP 49: 🚗 FULL MODEL DEPLOYMENT PIPELINE ON STREAMLIT FOR GRADIENT BOOST

# ✅ Streamlit provides a lightweight, interactive interface for deploying ML models.
# ✅ By integrating the serialized Gradient Boosting pipeline, stakeholders can test predictions
# in real time without needing to retrain or understand the underlying code.
# ✅ This step bridges research and production, ensuring reproducibility and accessibility

import streamlit as st
import joblib
import pandas as pd
import os
import tempfile

# 🔰 ATTEMPT TO LOAD THE TRAINED PIPELINE FROM COMMON LOCATIONS
possible_paths = [
    "gradient_boosting_pipeline.pkl",
    os.path.join("models", "gradient_boosting_pipeline.pkl"),
    os.path.join("saved_models", "gradient_boosting_pipeline.pkl")
]

model = None
for p in possible_paths:
    if os.path.exists(p):
        try:
            model = joblib.load(p)
            st.info(f"Loaded model from: {p}")
            break
        except Exception as e:
            st.warning(f"Found file at {p} but failed to load it: {e}")

# If model still not loaded, allow user to upload the .pkl file
if model is None:
    st.warning("Model file 'gradient_boosting_pipeline.pkl' not found in the app directory.")
    uploaded_file = st.file_uploader("Upload 'gradient_boosting_pipeline.pkl' (required to make predictions)", type=["pkl"])
    if uploaded_file is not None:
        # Save uploaded file to a temporary file and load with joblib
        with tempfile.NamedTemporaryFile(delete=False, suffix=".pkl") as tmp:
            tmp.write(uploaded_file.getvalue())
            tmp_path = tmp.name
        try:
            model = joblib.load(tmp_path)
            st.success("Model uploaded and loaded successfully.")
        except Exception as e:
            st.error(f"Failed to load uploaded model: {e}")
            st.stop()
        finally:
            try:
                os.remove(tmp_path)
            except Exception:
                pass
    else:
        st.error("Model is required. Please upload the 'gradient_boosting_pipeline.pkl' file or place it in the app directory.")
        st.stop()

# 🔰 STREAMLIT APP STRUCTURE
st.title("🚗 Car Price Prediction App (Gradient Boosting)")
st.write("This app uses a tuned Gradient Boosting model to predict car resale prices.")

# ✅ Input fields for numeric features
max_power = st.number_input("Max Power (bhp)", min_value=40.0, max_value=250.0, value=100.0, format="%.1f")
engine = st.number_input("Engine Size (cc)", min_value=800.0, max_value=4000.0, value=1500.0, format="%.1f")
mileage = st.number_input("Mileage (km/l)", min_value=5.0, max_value=25.0, value=15.0, format="%.2f")
vehicle_age = st.number_input("Vehicle Age (years)", min_value=0, max_value=20, value=5, step=1)

# 🔰 PREDICTION
if st.button("Predict Selling Price"):
    input_data = pd.DataFrame({
        'max_power': [float(max_power)],
        'engine': [float(engine)],
        'mileage': [float(mileage)],
        'vehicle_age': [int(vehicle_age)]
    })
    try:
        prediction = model.predict(input_data)
        st.success(f"✅ Predicted Selling Price: R{prediction[0]:,.2f}")
    except Exception as e:
        st.error(f"Prediction failed: {e}")
```
